```python
import pandas as pd
import shutil
import os

# Steps to create the Frontal Face Composition dataset (FFC) - 10k (10.000 imagens)

def create_ffc_10k_dataset(original_dataset_path, csv_file_path, destination_path):
    """
    Reorganizes image files based on a CSV mapping.

    Args:
        original_dataset_path (str): The path to the root directory of the image dataset.
        csv_file_path (str): The path to the CSV file with the mapping information.
        destination_path (str): The path to the directory where the reorganized images will be saved.
    """
    try:
        # Check if the image dataset path exists
        if not os.path.isdir(original_dataset_path):
            print(f"Erro: The image dataset was not found in '{original_dataset_path}'.")
            return

        # Check if the CSV file exists
        if not os.path.isfile(csv_file_path):
            print(f"Erro: The CSV file was not found in '{csv_file_path}'.")
            return

        # Check if the destination path exists, if not, create it
        if not os.path.exists(destination_path):
            print(f"Creating the target directory in '{destination_path}'...")
            os.makedirs(destination_path)

        # Read the CSV file into a pandas DataFrame
        df = pd.read_csv(csv_file_path)

        # Iterate over each row in the DataFrame
        for index, row in df.iterrows():
            original_dataset_name = row['dataset']
            original_class_name = row['original_class_name']
            original_image_name = row['original_image_name']
            filetype = row['filetype']
            new_class_name = row['new_class_name']
            new_image_name = row['new_image_name']

            # Construct the full path of the original image
            source_file = os.path.join(original_dataset_path, original_dataset_name, original_class_name, f"{original_image_name}.{filetype}")

            # Construct the path for the new subdirectory
            new_class_dir = os.path.join(destination_path, new_class_name)

            # Create the new subdirectory if it doesn't exist
            if not os.path.exists(new_class_dir):
                os.makedirs(new_class_dir)

            # Construct the full path for the destination image
            destination_file = os.path.join(new_class_dir, f"{new_image_name}-{original_image_name}.{filetype}")

            # Check if the source file exists before trying to copy
            if os.path.isfile(source_file):
                # Copy the file to the new location with the new name
                shutil.copy(source_file, destination_file)
                # print(f"Copied: '{source_file}' to '{destination_file}'")
            else:
                print(f"Warning: Source file not found: '{source_file}'. Skipping...")

        print("\nImage reorganization process completed successfully!")

    except KeyError as e:
        print(f"error: Column not found in CSV file. Verify that columns 'original_class_name', 'original_image_name', 'filetype', 'new_class_name', and 'new_image_name' exist. Error details: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```
